# 6. Lazy Evaluation and DAG (Lineage Graph)

Apache Spark follows the principle of **Lazy Evaluation**, meaning that transformations
are not executed immediately. Instead, Spark records all transformations and builds
a **Directed Acyclic Graph (DAG)**, also known as the **Lineage Graph**.

Execution begins only when an **Action** (such as `show()`, `count()`, or `collect()`)
is called.

This approach allows Spark to optimize the execution plan, reduce unnecessary
computations, and improve overall performance.

```python
# Transformation (No execution happens here)
filtered_df = df.filter(df.Region == "West") \
                .select("Order ID", "Customer Name", "Sales")

print("Transformation created successfully!")
```

### Observation

The previous operation is a **transformation**. Since no action has been performed,
Spark has not executed the query.

Instead, Spark records the sequence of operations in its execution plan (DAG)
and waits until an action is invoked.

```python
filtered_df.show(5)
```

### DAG Execution

When the `show()` action is called, Spark executes all pending transformations
defined in the DAG.

Execution Flow:

Read CSV
→ Filter Region = "West"
→ Select Required Columns
→ Display First 5 Records

This optimization is possible because Spark postpones execution until an action
is required.

```python
filtered_df.explain()
```

# 8. DataFrame Transformations

DataFrame transformations are operations that create a new DataFrame without
modifying the original one. These transformations are evaluated lazily and are
executed only when an action is called.

In this section, we will perform the following operations:

- Rename columns
- Convert data types
- Add a new calculated column

These transformations prepare the dataset for further analysis and processing.

## 8.1 Rename Columns

The `withColumnRenamed()` function is used to rename existing columns in a
DataFrame.

Renaming columns improves readability and helps maintain consistent naming
conventions.

```python
# Rename columns

df_renamed = df.withColumnRenamed("Customer Name", "Customer_Name") \
               .withColumnRenamed("Product Name", "Product_Name")

df_renamed.printSchema()
```

## 8.2 Cast Data Types

Some columns such as `Sales`, `Quantity`, and `Discount` were inferred as
string data types.

To perform mathematical calculations and aggregations, these columns need to
be converted into appropriate numeric data types using the `cast()` function.

```python
from pyspark.sql.functions import regexp_replace

df_casted = df_renamed \
    .withColumn(
        "Sales",
        regexp_replace("Sales", ",", "").cast("double")
    ) \
    .withColumn(
        "Quantity",
        regexp_replace("Quantity", ",", "").cast("int")
    ) \
    .withColumn(
        "Discount",
        regexp_replace("Discount", ",", "").cast("double")
    )

df_casted.printSchema()
```

```python
print(df_casted.dtypes)
```

## 8.3 Add New Column

The `withColumn()` function can also be used to create new columns based on
existing values.

Here, a new column named `Final_Sales` is created by adding 18% to the original
sales amount. This demonstrates how derived values can be calculated within a
Spark DataFrame.## 8.3 Add New Column

The `withColumn()` function can also be used to create new columns based on
existing values.

Here, a new column named `Final_Sales` is created by adding 18% to the original
sales amount. This demonstrates how derived values can be calculated within a
Spark DataFrame.

```python
from pyspark.sql.functions import col, round

df_transformed = df_casted.withColumn(
    "Final_Sales",
    round(col("Sales") * 1.18, 2)
)

df_transformed.select(
    "Sales",
    "Final_Sales"
).show(5)
```

# 9. Filtering and Selecting Data

Filtering and selecting are among the most commonly used DataFrame operations in
Apache Spark.

- **Filtering** is used to retrieve rows that satisfy a given condition.
- **Selecting** is used to retrieve only the required columns.

These operations reduce the amount of data processed and improve the efficiency
of Spark applications.

## 9.1 Selecting Columns

The `select()` function is used to retrieve only the required columns from a
DataFrame.

Selecting only the necessary columns reduces memory usage and improves query
performance.

```python
# Select a few important columns

selected_df = df_transformed.select(
    "Order ID",
    "Customer_Name",
    "Category",
    "Sales",
    "Final_Sales"
)

selected_df.show(5)
```

## 9.2 Filtering Rows

The `filter()` function is used to retrieve rows that satisfy a specified
condition.

In this example, only records belonging to the **Technology** category are
selected.

```python
technology_df = df_transformed.filter(
    df_transformed.Category == "Technology"
)

technology_df.show(5)
```

## 9.3 Filtering with Multiple Conditions

Spark allows multiple filtering conditions to be combined using logical
operators such as `AND (&)` and `OR (|)`.

The following example retrieves products that belong to the Technology
category and have sales greater than 500.

```python
filtered_df = df_transformed.filter(
    (df_transformed.Category == "Technology") &
    (df_transformed.Sales > 500)
)

filtered_df.select(
    "Product_Name",
    "Category",
    "Sales"
).show(5)
```

# 10. Handling Null Values

Handling missing or null values is an important step in data preprocessing.

Apache Spark provides built-in functions to:

- Identify null values
- Remove rows containing null values
- Replace null values with default values

Before performing any cleaning operations, it is good practice to inspect the dataset for missing values.

```python
from pyspark.sql.functions import col, when, count

null_counts = df_transformed.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in df_transformed.columns
])

null_counts.show(truncate=False)
```

### Observation

The null value count shows the number of missing values in each column.

For the Sample Superstore dataset, most or all columns are expected to contain no null values. Even when a dataset has no missing values, performing this check is considered a good data quality practice before starting analysis.

```python
df_no_null = df_transformed.dropna()

print("Rows before removing null values :", df_transformed.count())
print("Rows after removing null values  :", df_no_null.count())
```

```python
df_filled = df_transformed.fillna({
    "Customer_Name": "Unknown",
    "Sales": 0.0,
    "Discount": 0.0
})

df_filled.show(5)
```

# 11. Transformations and Actions

Apache Spark operations are classified into two categories:

### Transformations
Transformations create a new DataFrame from an existing one without executing
the computation immediately. Spark records these operations and executes them
only when an action is called. This behavior is known as **Lazy Evaluation**.

Examples:
- `select()`
- `filter()`
- `withColumn()`
- `withColumnRenamed()`
- `drop()`

### Actions
Actions trigger the execution of all pending transformations and return results
or write data to storage.

Examples:
- `show()`
- `count()`
- `collect()`
- `write()`

Understanding the difference between transformations and actions helps in writing
efficient Spark applications.

## 11.1 Example of a Transformation

The following operation filters the dataset to include only products from the
Technology category.

Since no action is performed, Spark only records the transformation and does
not execute it immediately.

```python
technology_df = df_transformed.filter(
    df_transformed.Category == "Technology"
)
```

## 11.2 Example of an Action

The `show()` function is an action.

When it is called, Spark executes all pending transformations and displays the
result.

```python
technology_df.show(5)
```

```python
print("Total Technology Products:", technology_df.count())
```

### Observation

The `filter()` operation alone does not execute immediately because it is a
transformation.

The execution starts only when an action such as `show()` or `count()` is called.
This optimization is one of the key reasons why Apache Spark performs efficiently
on large datasets.

# 12. Wide Transformations and Performance Concepts

Apache Spark performs data processing using two types of transformations:

- **Narrow Transformations**
- **Wide Transformations**

Understanding these transformations is important because they directly affect
Spark's performance.

Wide transformations involve **data shuffling** across partitions, making them
more expensive than narrow transformations.

Spark also improves performance using optimization techniques such as
**Predicate Pushdown**, which reduces the amount of data read from storage.

## 12.1 Narrow Transformation

A narrow transformation is one in which each output partition depends on only
one input partition.

These transformations do not require data movement between executors, making
them faster and more efficient.

Examples include:

- select()
- filter()
- withColumn()

```python
narrow_df = df_transformed.filter(
    df_transformed.Region == "West"
)

narrow_df.show(5)
```

## 12.2 Wide Transformation

A wide transformation requires data to be redistributed across partitions.

This process is known as **Shuffle**.

Wide transformations are generally more expensive because Spark needs to
exchange data between executors before producing the final result.

Common examples include:

- groupBy()
- join()
- distinct()
- orderBy()

```python
sales_by_region = df_transformed.groupBy("Region") \
                                .sum("Sales")

sales_by_region.show()
```

## 12.3 Shuffle

Shuffle is the process of redistributing data across partitions during wide
transformations.

Although shuffle enables operations such as grouping and joining, it increases:

- Disk I/O
- Network communication
- Execution time

Reducing unnecessary shuffle operations is considered a Spark performance best
practice.

## 12.4 Predicate Pushdown

Predicate Pushdown is a Spark optimization technique where filter conditions
are pushed down to the data source.

Instead of reading the entire dataset into memory and then filtering it, Spark
reads only the rows that satisfy the filter condition.

This optimization significantly improves performance, especially when working
with columnar formats such as Parquet.

```python
west_orders = df_transformed.filter(
    df_transformed.Region == "West"
)

west_orders.show(5)
```

# 13. Building a Spark Data Pipeline

A data pipeline is a sequence of data processing steps that transforms raw data
into a format suitable for analysis or storage.

In this project, the pipeline performs the following operations:

1. Read the input CSV file.
2. Apply transformations.
3. Filter the required data.
4. Save the processed data in CSV format.
5. Save the processed data in Parquet format.

This demonstrates a simple ETL (Extract, Transform, Load) workflow using Apache Spark.

```python
# Create a processed DataFrame

pipeline_df = df_transformed.filter(
    df_transformed.Category == "Technology"
).select(
    "Order ID",
    "Customer_Name",
    "Category",
    "Sales",
    "Final_Sales",
    "Profit"
)

pipeline_df.show(5)
```

## 13.1 Writing Data to CSV

The processed dataset is written to a CSV file.

The `overwrite` mode replaces any existing output folder with the latest results.

```python
pipeline_df.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("output/csv")
```

## 13.2 Writing Data to Parquet

Parquet is a columnar storage format optimized for analytical workloads.

Compared to CSV, Parquet offers:

- Better compression
- Faster query performance
- Efficient storage
- Support for Predicate Pushdown

```python
pipeline_df.write \
    .mode("overwrite") \
    .parquet("output/parquet")
```

## 13.3 Reading the Parquet File

The saved Parquet file is read back to verify that the write operation was
successful.

```python
parquet_df = spark.read.parquet("output/parquet")

parquet_df.show(5)
```

## 13.4 CSV vs Parquet

CSV and Parquet are commonly used file formats in data engineering.

| CSV | Parquet |
|------|----------|
| Row-based storage | Column-based storage |
| Larger file size | Compressed storage |
| Slower analytical queries | Faster analytical queries |
| No Predicate Pushdown | Supports Predicate Pushdown |
| Plain text format | Binary format |

# 14. Best Practices for Large Datasets

When working with large datasets in Apache Spark, following best practices helps
improve performance and resource utilization.

Some recommended practices include:

- Use `show()` instead of `collect()` whenever possible.
- Filter data early to reduce processing.
- Select only the required columns.
- Prefer Parquet over CSV for analytical workloads.
- Avoid unnecessary shuffle operations.
- Cache intermediate DataFrames only when reused multiple times.

```python
# Display only the first five rows

pipeline_df.show(5)
```

### Why avoid `collect()`?

The `collect()` function transfers the entire DataFrame from the Spark cluster
to the driver program.

For large datasets, this can consume excessive memory and may cause the driver
application to fail.

The `show()` function is preferred because it retrieves only a limited number of
rows for inspection.
